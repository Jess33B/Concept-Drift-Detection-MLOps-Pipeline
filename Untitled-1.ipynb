```python
import pandas as pd

# Load both
train_df = pd.read_csv("C:\\Users\\JESNABINU\\Downloads\\Airplane\\train.csv")
test_df = pd.read_csv("C:\\Users\\JESNABINU\\Downloads\\Airplane\\test.csv")

# Drop columns
train_df = train_df.drop(columns=["Unnamed: 0", "id"], errors="ignore")
test_df = test_df.drop(columns=["Unnamed: 0", "id"], errors="ignore")

# Drop missing
train_df = train_df.dropna()
test_df = test_df.dropna()

# Convert target
mapping = {
    'satisfied': 1,
    'neutral or dissatisfied': 0
}
train_df['satisfaction'] = train_df['satisfaction'].map(mapping)
test_df['satisfaction'] = test_df['satisfaction'].map(mapping)

# ✅ COMBINE
combined = pd.concat([train_df, test_df], axis=0)

# ✅ ENCODE ONCE
combined = pd.get_dummies(combined, drop_first=True)

# ✅ SPLIT BACK
train_df = combined.iloc[:len(train_df)]
test_df = combined.iloc[len(train_df):]
```

```python
X_train = train_df.drop("satisfaction", axis=1)
X_test = test_df.drop("satisfaction", axis=1)
```

```python
X_train = train_df.drop("satisfaction", axis=1)
X_test = test_df.drop("satisfaction", axis=1)
```

```python
from scipy.stats import ks_2samp

drift_results = {}

for col in X_train.columns:
    stat, p_value = ks_2samp(X_train[col], X_test[col])
    
    drift_results[col] = {
        "p_value": p_value,
        "drift": p_value < 0.05
    }
```

```python
import pandas as pd

drift_df = pd.DataFrame(drift_results).T
print(drift_df)
```

```python
drifted_features = drift_df[drift_df["drift"] == True]

print("Drifted features:", len(drifted_features))
print(drifted_features)
if len(drifted_features) > 0:
    print("⚠️ Drift detected")
else:
    print("✅ No drift")
```

```python
from sklearn.ensemble import RandomForestClassifier

y_test = test_df["satisfaction"]

if len(drifted_features) > 0:
    new_model = RandomForestClassifier()
    new_model.fit(X_test, y_test)
```
